```python
## Handling null values and Data Preprocesssing

c = ["Applicant_Income", "Coapplicant_Income", "Age", "Dependents", "Credit_Score", "Existing_Loans", "DTI_Ratio", "Savings", "Collateral_Value", "Loan_Amount", "Loan_Term"]

for col in c:
    df[col] = df[col].fillna(df[col].median())

columns = list(df.columns)
c2 = [col for col in columns if col not in c]
c2.remove('Applicant_ID')
c2.remove('Loan_Approved')

for col in c2:
    df[col] = df[col].fillna(df[col].mode()[0])

# df["Applicant_Income"] = df["Applicant_Income"].fillna(df["Applicant_Income"].median())
# df["Coapplicant_Income"] = df["Coapplicant_Income"].fillna(df["Coapplicant_Income"].median())
# df["Age"] = df["Age"].fillna(df["Age"].median())
# df["Dependents"] = df["Dependents"].fillna(df["Dependents"].median())
# df["Credit_Score"] = df["Credit_Score"].fillna(df["Credit_Score"].median())
df = df.drop(columns = ["Applicant_ID"])

df["Gender"] = df["Gender"].map({
    "Male":1,
    "Female":0
})

df["Loan_Approved"] = df["Loan_Approved"].map({
    "Yes" : 1,
    "No" : 0
})

df
```

```python
## Train Test Split

untested_data = df[df["Loan_Approved"].isna()]

data = df.dropna()

X = data.drop(columns = ["Loan_Approved"])
y = data["Loan_Approved"]

## Train Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.2, random_state = 18)
```

```python
## One Hot Encoding
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown = "ignore")

X_train_numerical = X_train.drop(columns = c2)
X_test_numerical = X_test.drop(columns = c2)

X_train_encoded = encoder.fit_transform(X_train[c2]).toarray()
X_test_encoded = encoder.transform(X_test[c2]).toarray()

encoded_columns = encoder.get_feature_names_out(c2)

X_train_categorical = pd.DataFrame(
    X_train_encoded,
    columns = encoded_columns,
    index = X_train.index
)

X_test_categorical = pd.DataFrame(
    X_test_encoded,
    columns = encoded_columns,
    index = X_test.index
)

## Concatenation
X_train_encoded = pd.concat(
    [X_train_numerical, X_train_categorical],
    axis = 1
)

X_test_encoded = pd.concat(
    [X_test_numerical, X_test_categorical],
    axis = 1
)
```

```python
## Scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_encoded)
X_test_scaled = scaler.transform(X_test_encoded)
```

```python
## Model - kNN

from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors = 19)
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)
```

```python
## Evaluation

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Accuracy : {accuracy}    Precision : {precision}    Recall : {recall}    F1-score : {f1}")
```

```python
## Fine Tuning the kNN model using different values of n_neighbors (k = 11 appears to be the best choice)

for k in range(1,32,2):
    model = KNeighborsClassifier(n_neighbors = k)
    model.fit(X_train_scaled, y_train)
    
    y_pred = model.predict(X_test_scaled)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    
    print(f"For k = {k} ->  Accuracy : {accuracy}    Precision : {precision}    Recall : {recall}    F1-score : {f1}")
```

```python
## Model - Logistic Regression

from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"For Logistic Regression ->  Accuracy : {accuracy}    Precision : {precision}    Recall : {recall}    F1-score : {f1}")
```

```python
## Model - Gaussian NB

from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"For Gaussian Naive Bayes ->  Accuracy : {accuracy}    Precision : {precision}    Recall : {recall}    F1-score : {f1}")
```
